```python
import get_wa_scores as wa
import util

data = util.load_data()

# Add column "wa_score" with None entries
data["wa_score"] = None

data
```

```python

MAPPING = {
    # Laufdisziplinen
    "100 m": "100m",
    "200 m": "200m",
    "400 m": "400m",
    "800 m": "800m",
    "1500 m": "1500m",
    "3000 m": "3000m",
    "5000 m": "5000m",
    "10 km": "10km",
    "Halbmarathon": "HM",
    "Marathon": "Marathon",
    "100 km": "100km",
    "300 m": "300m",
    "1000 m": "1000m",
    "5 km": "5km",
    "10000 m": "10000m",

    # Hindernis (→ mSC)
    "3000 m Hindernis": "3000mSC",
    "2000 m Hindernis": "2000mSC",
    # not in the list: "1500 m Hindernis": "1500mSC",

    # Hürden (→ mH)
    "100 m Huerden": "100mH",
    "110 m Huerden": "110mH",
    "400 m Huerden": "400mH",

    # Gehen (→ mW / kmW)
    "3000 m Gehen": "3000mW",
    "10000 m Gehen": "10000mW",
    "50 km Gehen": "50kmW",

    # Sprung/Wurf (Meterdisziplinen → Kürzel)
    "Hochsprung": "HJ",
    "Stabhochsprung": "PV",
    "Weitsprung": "LJ",
    "Dreisprung": "TJ",
    "Kugelstoss": "SP",
    "Diskuswurf": "DT",
    "Hammerwurf": "HT",
    "Speerwurf": "JT",

    # Mehrkampf
    "Siebenkampf": "Heptathlon",
    "Zehnkampf": "Decathlon"
}

for index, row in data.iterrows():
    discipline = row["disziplin"]
    if discipline in MAPPING:
        discipline_code = MAPPING[discipline]
        performance = row["leistung"]
        gender = row["geschlecht"]
        if gender == "W":
            gender = "female"
        else:
            gender = "male"
        wa_score = wa.calculate_event_points(gender, discipline_code, seconds=performance)
        points = wa_score["points"]
        data.at[index, "wa_score"] = points
    else:
        data.at[index, "wa_score"]
        print(f"Discipline '{discipline}' not found in mapping.")

data.to_csv("data_with_wa_scores.csv", index=False)
```
