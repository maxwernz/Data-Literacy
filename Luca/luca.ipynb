```python
import pdfplumber
import pandas as pd
import re
import matplotlib.pyplot as plt
```

```python


# --- 1. Datei einlesen ---
# Wir definieren den Dateinamen
dateiname = 'list01-24_all.csv'

try:
    # Wir lesen die CSV-Datei ein. 
    # WICHTIG: Wir sagen pandas, dass das Trennzeichen ein Semikolon (;) ist.
    df = pd.read_csv(dateiname, sep=';')

    print(f"Datei '{dateiname}' erfolgreich geladen. Starte Analyse...")

    # --- 2. Nach '100 m' filtern ---
    # Wir erstellen eine neue, kleinere Tabelle (DataFrame),
    # die NUR die Zeilen enthält, bei denen in der Spalte 'disziplin' der Wert '100 m' steht.
    df_100m = df[df['disziplin'] == '100 m'].copy() # .copy() ist saubere Praxis

    # --- 3. Leistung in eine Zahl umwandeln (der wichtigste Schritt) ---
    # Die Spalte 'leistung' ist Text (z.B. "9,99"). 
    # Um den kleinsten Wert zu finden, braucht Python eine Zahl (z.B. 9.99).
    
    # Wir erstellen eine NEUE Spalte 'leistung_sekunden'
    # 1. Wir nehmen die 'leistung'-Spalte
    # 2. Ersetzen das Komma (',') durch einen Punkt ('.')
    # 3. Wandeln diesen Text in eine Zahl (float) um.
    #    pd.to_numeric ist robust: errors='coerce' wandelt fehlerhafte
    #    Einträge (z.B. "n.a.") einfach in 'NaN' (Not a Number) um, 
    #    ohne das Programm abstürzen zu lassen.
    df_100m['leistung_sekunden'] = pd.to_numeric(
        df_100m['leistung'].str.replace(',', '.'),
        errors='coerce'
    )

    # --- 4. Schnellste Zeit finden ---
    # .min() findet einfach den kleinsten Wert in unserer neuen Zahlenspalte.
    schnellste_zeit = df_100m['leistung_sekunden'].min()

    # --- 5. Den Eintrag zu dieser Zeit finden ---
    # Jetzt filtern wir unsere 100m-Tabelle: "Zeige mir alle Zeilen, 
    # wo die 'leistung_sekunden' gleich unserer 'schnellste_zeit' ist."
    # (Es könnten ja mehrere Personen die gleiche Zeit gelaufen sein)
    schnellste_person_info = df_100m[df_100m['leistung_sekunden'] == schnellste_zeit]

    # --- 6. Ergebnis ausgeben ---
    print(f"\nDie schnellste gefundene Zeit für '100 m' war: {schnellste_zeit} Sekunden.")
    print("\n--- Details zum Datensatz ---")
    
    # .to_string(index=False) sorgt für eine saubere Ausgabe 
    # ohne die pandas-Zeilennummern (den Index).
    print(schnellste_person_info.to_string(index=False))


except FileNotFoundError:
    print(f"FEHLER: Die Datei '{dateiname}' wurde nicht gefunden.")
    print("Stellen Sie sicher, dass die Datei im selben Ordner wie das Skript liegt.")
except Exception as e:
    print(f"Ein unerwarteter Fehler ist aufgetreten: {e}")
```
