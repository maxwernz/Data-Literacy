```python
import numpy as np
# import matplotlib.pyplot as plt

x = np.linspace(0, np.pi, 20)
offsets = np.linspace(0, 2 * np.pi, 8, endpoint=False)
yy = [np.sin(x + phi) for phi in offsets]
for y in yy:
    plt.plot(x, y, label=f"offset {y[0]:.2f}")
plt.legend()
plt.grid()
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("Test Plotting Style")
plt.savefig("plots/test_plotting_style")
plt.show()
```

```python
rng = np.random.default_rng(42)
x = rng.normal(size=200)
y = x + rng.normal(scale=0.5, size=200)

plt.figure()
plt.scatter(x, y, alpha=0.6)
plt.xlabel("x")
plt.ylabel("y")
```

```python
data = rng.normal(size=1_000)

plt.figure()
plt.hist(data, bins=40, density=True, alpha=0.6)

# einfache Glättung
from scipy.stats import gaussian_kde
xs = np.linspace(data.min(), data.max(), 300)
plt.plot(xs, gaussian_kde(data)(xs))

plt.xlabel("Wert")
plt.ylabel("Dichte")
```

```python
plt.figure()

for i in range(5):
    plt.plot(x, np.sin(x + i * 0.3), label=f"shift={i}")

plt.xlabel("x")
plt.ylabel("Amplitude")
plt.legend(ncol=2)
```

```python
x = np.arange(5)
y = rng.normal(loc=1.0, scale=0.2, size=5)
yerr = rng.uniform(0.05, 0.15, size=5)

plt.figure()
plt.errorbar(x, y, yerr=yerr, fmt="o", capsize=3)
plt.xlabel("Bedingung")
plt.ylabel("Messwert")
```

```python
data = [
    rng.normal(0, 1, 200),
    rng.normal(0.5, 1, 200),
    rng.normal(1.0, 1, 200),
]

plt.figure()
plt.boxplot(data, tick_labels=["A", "B", "C"])
plt.ylabel("Wert")
```

```python
corr = rng.normal(size=(10, 10))
corr = (corr + corr.T) / 2

plt.figure()
im = plt.imshow(corr, aspect="auto")
plt.colorbar(im, fraction=0.046)
plt.title("Heatmap")
```
