```python
def parse_pdf(pdf_file):

    decathlon_patterns =  re.compile(
        r"("
        r"^(Zehnkampf|10-Kampf)"
        r")"
    )

    with pdfplumber.open(pdf_file) as pdf:
        for page in pdf.pages:
            text = page.extract_text()
            if not text:
                continue
            for line in text.split("\n"):
                line = line.strip()
                if not line:
                    continue

                decathlon_found = decathlon_patterns.match(line)
                if decathlon_found and "Mannschaft" not in line:
                    print(f"Found header line: {line}")

                

                # if not line.startswith("Zehkampf") and not line.startswith("10-Kampf"):
                #     continue

                # print(line)
                # break
```

```python
pdf_year = ["data01-17", "data18-22", "data24", "data23"]
 
decathlon_events = {}

for year in pdf_year:
    pdf_folder = Path(f"../../Data_pdf/{year}")
    pdf_files = list(pdf_folder.glob("*.pdf"))
    
    print(f"\n📂 Verarbeite Ordner: {pdf_folder} ({len(pdf_files)} Dateien)")
    
    for pdf_file in tqdm(pdf_files, desc=f"{year}", unit="pdf"):
        decathlon_events[pdf_file.stem] = [parse_pdf(pdf_file)]
```
