# IAAF Points Fairness
- are the iaaf-points fair across different disciplines

## Basic ides
- Discipline A: small improvement -> lot of points
- Discipline B: same improvement -> less points

## Fairness definiton
Same relative improvement -> same points

$\text{Fairness} = \dfrac {\Delta \text{IAAF} - \text{Points}}{\Delta \text{Perfomance} / \text{Perfomance}}$

-> How much points do 1% perfomance improvement gain

```python
running_disciplines = util.MEASUREMENT_TYPE['time']

# Align performance such that higher is better for all disciplines
data['perf_aligned'] = data['leistung']
data.loc[data['disziplin'].isin(running_disciplines), 'perf_aligned'] *= -1
```

```python
def compute_local_fairness(group):
    group = group.sort_values('perf_aligned')

    group['d_points'] = group['iaaf_score'].diff()
    group['d_perf'] = group['perf_aligned'].diff()

    group['rel_improvement'] = group['d_perf'] / group['perf_aligned'].shift(1)
    group['points_per_percent'] = group['d_points'] / group['rel_improvement']

    return group
```

```python
df_fair = (data
             .groupby('disziplin')
             .apply(compute_local_fairness)
            )
```

```python
df_fair = df_fair.replace([np.inf, -np.inf], np.nan)
df_fair = df_fair.dropna(subset=['points_per_percent'])

lower = df_fair['points_per_percent'].quantile(0.05)
upper = df_fair['points_per_percent'].quantile(0.95)

df_fair = df_fair[
    (df_fair['points_per_percent'] >= lower) &
    (df_fair['points_per_percent'] <= upper)
]
```

```python
df_fair = df_fair.reset_index(drop=True)

fairness_summary = (
    df_fair
        .groupby('disziplin')['points_per_percent']
        .agg(['median', 'mean', 'std', 'count'])
        .sort_values('median', ascending=False)
)
```

```python
fairness_summary    
```

```python
plt.figure(figsize=(12, 6))

df_fair.boxplot(column='points_per_percent', by='disziplin', rot=45)
plt.title('IAAF-Fairness: Points per 1% Improvement by Discipline')
plt.suptitle('')
plt.ylabel('Points per 1% Improvement')
plt.tight_layout()
```
