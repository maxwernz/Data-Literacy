## Bootstrap estimates

Can girls run faster than boys? To approach an answer, we first have to be more precise, and translate the question into a more formal statement: There are two probability *distributions* here $p(Time \mid Sex, Age)$ (we keep the age fixed). Do we mean the *fastest* girl runs faster than the fastest boy? The *average* girl? Or something about the *spread* of the distribution? For each of these quantities, we can build bootstrap estimates:

```python
global NBOOT, N
NBOOT = 10000 # number of bootstrap replicates to build
N = 100 # np.min([n_f, n_m]) # let's fix the size of each group to be equal, to avoid problems with the varying number of runners

Simulation_girls = np.random.choice(df_f["finish_time"], size = (NBOOT, N), replace = True) # Nboot x N matrix of bootstrap samples
Simulation_boys = np.random.choice(df_m["finish_time"], size = (NBOOT, N), replace = True) # Nboot x N matrix of bootstrap samples

Fastest_girls = np.min(Simulation_girls, axis = 1) # fastest time for each bootstrap sample
Fastest_boys = np.min(Simulation_boys, axis = 1) # fastest time for eäach bootstrap sample

# compute the difference in fastest times between the two groups
Diff = Fastest_girls - Fastest_boys
average_of_difference = (Simulation_girls - Simulation_boys).mean(axis = 1)
# note that this is the same as:
# difference_of_averages = np.mean(Simulation_girls, axis = 1) - np.mean(Simulation_boys, axis = 1) 

Mean_girls = np.mean(Simulation_girls, axis = 1)
Mean_boys = np.mean(Simulation_boys, axis=1)

Std_girls = np.std(Simulation_girls, axis = 1)
Std_boys = np.std(Simulation_boys, axis = 1)
```

```python
fig, ax = plt.subplots()
ax.set_title(f"{NBOOT} bootstrap samples of {N} runners")
ax.set_xlabel("Time (gross)")

np.random.seed(1)
u_g = np.random.rand(N)
u_b = np.random.rand(N)

for i in np.arange(10):
    ax.plot(Simulation_girls[i,:], 0.1 + 0.3 * u_g + i, "o", color=rgb.tue_red, alpha=0.5, mec = "none", ms = 2)
    ax.plot(Simulation_boys[i,:], 0.6 + 0.3 * u_b + i, "o", color=rgb.tue_blue, alpha=0.5, mec = "none", ms = 2)
    ax.axhline(i, color = rgb.tue_dark, alpha = 0.5)
    ax.axhline(i + 0.5, color = rgb.tue_dark, alpha = 0.25, ls = '--')
ax.axhline(10, color = rgb.tue_dark, alpha = 0.5)

# fastest girls and boys (front of the distribution)
counts, bins = np.histogram(Fastest_girls, bins=int(np.sqrt(NBOOT)))
ax.bar(bins[:-1],counts / np.max(counts), bottom=10, width = bins[1] - bins[0], alpha = 1.0, color = rgb.tue_red)
counts, bins = np.histogram(Fastest_boys, bins=int(np.sqrt(NBOOT)))
ax.bar(bins[:-1],counts / np.max(counts), bottom=10, width = bins[1] - bins[0], alpha = 1.0, color = rgb.tue_blue)

# mean runtime
counts, bins = np.histogram(Mean_girls, bins=int(np.sqrt(NBOOT)))
ax.bar(bins[:-1],counts / np.max(counts), bottom=10, width = bins[1] - bins[0], alpha = 0.5, color = rgb.tue_red)
counts, bins = np.histogram(Mean_boys, bins=int(np.sqrt(NBOOT)))
ax.bar(bins[:-1],counts / np.max(counts), bottom=10, width = bins[1] - bins[0], alpha = 0.5, color = rgb.tue_blue)

ax.xaxis.set_major_locator(plt.MultipleLocator(10))
ax.xaxis.set_minor_locator(plt.MultipleLocator(1))
ax.xaxis.set_major_formatter(plt.FuncFormatter(format_seconds))
ax.grid(axis = 'x', which = 'major', alpha = 0.5)
ax.set_yticks([])
ax.set_ylabel("Bootstrap sample")
ax.set_ylim(0.0, 11.25)

ax.axvline(df_f.iloc[63]["finish_time"], color = rgb.tue_red, alpha = 0.5)
ax.axvline(df_m.iloc[48]["finish_time"], color = rgb.tue_blue, alpha = 0.5)
ax.axvline(df_m.iloc[1]["finish_time"], color = rgb.tue_blue, alpha = 0.5)

fig.savefig("erbelauf_U10_bootstrap.pdf")
```

One way to interpret and use the results of the bootstrap is to "control for the effect of sex" by subtracting the difference between the means of the two groups from the girls' times. So that "after correction, the two groups have the same average runtime. 

```python
Delta = (Mean_girls - Mean_boys).mean(axis=0)
# 95% CI of the difference in means
CI = np.percentile(Mean_girls - Mean_boys, [2.5, 97.5]) - Delta

Corrected_Girls = df_f["finish_time"] - Delta
# the variance of the difference in means is the sum of the variances of the MEANS (not the groups themselves)
Error = np.sqrt((Mean_girls - Mean_boys).mean(axis=0))

fig, ax = plt.subplots()

ax.plot(
    df_f["finish_time"],
    0.5 * u_f,
    "o",
    label=None,
    color=rgb.tue_red,
    alpha=0.5,
    mec="none",
    ms=4,
)
ax.plot(
    df_m["finish_time"],
    1 + 0.5 * u_m,
    "o",
    label=None,
    color=rgb.tue_blue,
    alpha=0.5,
    mec="none",
    ms=4,
)

ax.errorbar(
    x=Corrected_Girls,
    y=2 + 0.5 * u_f,
    xerr=np.abs(np.zeros_like(Corrected_Girls) + CI[:,None]),
    fmt="o",
    label=f"{n_f} Girls `corrected`",
    color=rgb.tue_orange,
    alpha=0.5,
    mec="none",
    ms=4,
)

ax.plot(
    df_f.iloc[i_hennig]['finish_time'],
    0.5 * u_f[i_hennig],
    "o",
    label=df_f.iloc[i_hennig]['first_name'] + " " + df_f.iloc[i_hennig]['surname'],
    color=rgb.tue_red,
    alpha=1.0,
    mec="none",
)

ax.plot(
    df_m.iloc[i_hein]['finish_time'],
    1 + 0.5 * u_m[i_hein],
    "o",
    label=df_m.iloc[i_hein]['first_name'] + " " + df_m.iloc[i_hein]['surname'],
    color=rgb.tue_blue,
    alpha=1.0,
    mec="none",
)

ax.plot(
    df_m.iloc[i_macke]['finish_time'],
    1 + 0.5 * u_m[i_macke],
    "s",
    label=df_m.iloc[i_macke]['first_name'] + " " + df_m.iloc[i_macke]['surname'],
    color=rgb.tue_blue,
    alpha=1.0,
    mec="none",
)

ax.plot(
    Corrected_Girls[i_hennig],
    2 + 0.5 * u_f[i_hennig],
    "o",
    label=None,
    color=rgb.tue_orange,
    alpha=1.0,
    mec="none",
)

# set the x-axis formatter to the custom function
ax.xaxis.set_major_locator(plt.MultipleLocator(10))
ax.xaxis.set_minor_locator(plt.MultipleLocator(1))
ax.xaxis.set_major_formatter(plt.FuncFormatter(format_seconds))
ax.legend(loc=("lower left"), framealpha=1.0, facecolor="white", edgecolor="none")
ax.grid(axis="x", which="major", alpha=0.5)
ax.yaxis.set_visible(False)
ax.set_title("Corrected by average difference in means")
ax.axhline(0.75, color=rgb.tue_dark, alpha=0.5)
ax.axhline(1.75, color=rgb.tue_dark, alpha=0.5)
ax.set_ylim(-0.25, 2.75)

fig.savefig("erbelauf_U10_corrected.pdf")
```

We have to be very careful here about our choice of estimator. What is it that we want to estimate?

* If we randomly pick one girl and one boy from the population, what is the chance that the girl wins a 750m race against the boy?: $\mathbb{E}_{p}(\operatorname{sign}(t_f - t_m))$
* In a randomly selected group of $N$ boys and $N$ girls, what is the chance that *the fastest girl of the group* is faster than *the fastest boy in the group*? $\mathbb{E}_p(\operatorname{sign}(\min(t_f) - \min(t_g)))$
* Among the top $k$ runners, how many are girls?

The underlying problem here is that our confidence about the CDF of the distributions (which are *functions*) is not homogenous:

```python
# let's estimate the empirical CDF of each group across the bootstrap samples:
# Define the number of bins for the histogram
y = np.linspace(120, 280, 100)

lim = 500
# Initialize arrays to store the CDFs
cdf_boys = np.zeros((lim, len(y)))
cdf_girls = np.zeros((lim, len(y)))

# Compute the empirical CDF for each bootstrap sample
for i in tqdm(range(lim)):
    cdf_boys[i] = np.array([np.sum(Simulation_boys[i] <= yi) for yi in y]) / len(
        Simulation_boys[i]
    )
    cdf_girls[i] = np.array([np.sum(Simulation_girls[i] <= yi) for yi in y]) / len(
        Simulation_girls[i]
    )

fig, ax = plt.subplots()

for i in np.arange(lim):
    ax.plot(y, cdf_boys[i, :], alpha=0.03, color=rgb.tue_blue)
    ax.plot(y, cdf_girls[i, :], alpha=0.03, color=rgb.tue_red)

ticks = np.arange(120, 281, 20)
ax.set_xticks(ticks)
ax.set_xticklabels([format_seconds(i, 0) for i in ticks])
ax.set_ylabel("CDF($t$)")
ax.set_xlabel("Time $t$ (for 700m)")
ax.yaxis.set_major_locator(plt.MultipleLocator(0.1))
ax.yaxis.set_major_formatter(plt.FormatStrFormatter('%.1f'))


alpha = 0.05
DKW_f_lower = np.array(
    [
        np.max([0, cdf_f[i] - np.sqrt(np.log(2 / alpha) / (2 * n_f))])
        for i in range(len(y))
    ]
)
DKW_f_upper = np.array(
    [
        np.min([1, cdf_f[i] + np.sqrt(np.log(2 / alpha) / (2 * n_f))])
        for i in range(len(y))
    ]
)
DKW_m_lower = np.array(
    [
        np.max([0, cdf_m[i] - np.sqrt(np.log(2 / alpha) / (2 * n_m))])
        for i in range(len(y))
    ]
)
DKW_m_upper = np.array(
    [
        np.min([1, cdf_m[i] + np.sqrt(np.log(2 / alpha) / (2 * n_m))])
        for i in range(len(y))
    ]
)

ax.fill_between(
    y,
    DKW_f_lower,
    DKW_f_upper,
    color=rgb.tue_red,
    alpha=0.1,
    label=rf"DKW ${(1-alpha) * 100:.0f}\%$ confidence interval",
)
ax.fill_between(
    y,
    DKW_m_lower,
    DKW_m_upper,
    color=rgb.tue_blue,
    alpha=0.1,
    # label=fr"DKW $\alpha={alpha * 100:.0f}\%$ interval of boys",
)

ax.grid(axis="y", which="major", alpha=0.5)
ax.set_ylim(0, 1)
ax.set_xlim(120, 280)
ax.set_title(f"Empirical CDFs of {lim} bootstrap samples of {N} runners")
ax.legend(loc="lower right", framealpha=1.0, facecolor="white")

plt.savefig("erbelauf_U10_bootstrap3.pdf")
```

```python
# these define normal distributions over the running speeds in each age/sex group. If we assume that our observed groups are iid. samples from these distributions, we can draw simulated datasets and check how often "a random boy" would win against "a random girl" in each age group:

U10_f = df_f["finish_time"]
U10_m = df_m["finish_time"]

NBOOT = 10000

u_10_f_samples = np.random.normal(U10_f.mean(), U10_f.std(), size = (NBOOT,))
u_10_m_samples = np.random.normal(U10_m.mean(), U10_m.std(), size = (NBOOT,))
win_rate = (u_10_m_samples < u_10_f_samples).sum() / NBOOT
print(f"U10 female win rate: {win_rate:.2f}")

u_12_f_samples = np.random.normal(U12_f.mean(), U12_f.std(), size = (NBOOT,))
u_12_m_samples = np.random.normal(U12_m.mean(), U12_m.std(), size = (NBOOT,))
win_rate = (u_12_m_samples < u_12_f_samples).sum() / NBOOT
print(f"U12 female win rate: {win_rate:.2f}")

u_14_f_samples = np.random.normal(U14_f.mean(), U14_f.std(), size = (NBOOT,))
u_14_m_samples = np.random.normal(U14_m.mean(), U14_m.std(), size = (NBOOT,))
win_rate = (u_14_m_samples < u_14_f_samples).sum() / NBOOT
print(f"U14 female win rate: {win_rate:.2f}")

u_16_f_samples = np.random.normal(U16_f.mean(), U16_f.std(), size = (NBOOT,))
u_16_m_samples = np.random.normal(U16_m.mean(), U16_m.std(), size = (NBOOT,))
win_rate = (u_16_m_samples < u_16_f_samples).sum() / NBOOT
print(f"U16 female win rate: {win_rate:.2f}")
```

```python
# the alternative is a direct bootstrap:
NBOOT = 10000

u_10_f_samples = np.random.choice(U10_f, size = NBOOT, replace = True)
u_10_m_samples = np.random.choice(U10_m, size = NBOOT, replace = True)
win_rate = (u_10_m_samples > u_10_f_samples).sum() / NBOOT
print(f"For U10, a random girl is faster than a random boye {win_rate:.2f} of the time")

u_12_f_samples = np.random.choice(U12_f, size = NBOOT, replace = True)
u_12_m_samples = np.random.choice(U12_m, size = NBOOT, replace = True)
win_rate = (u_12_m_samples > u_12_f_samples).sum() / NBOOT
print(f"For U12, a random girl is faster than a random boye {win_rate:.2f} of the time")

u_14_f_samples = np.random.choice(U14_f, size = NBOOT, replace = True)
u_14_m_samples = np.random.choice(U14_m, size = NBOOT, replace = True)
win_rate = (u_14_m_samples > u_14_f_samples).sum() / NBOOT
print(f"For U14, a random girl is faster than a random boye {win_rate:.2f} of the time")

u_16_f_samples = np.random.choice(U16_f, size = NBOOT, replace = True)
u_16_m_samples = np.random.choice(U16_m, size = NBOOT, replace = True)
win_rate = (u_16_m_samples > u_16_f_samples).sum() / NBOOT
print(f"For U16, a random girl is faster than a random boye {win_rate:.2f} of the time")
```
