```python
# the following code downloads four files, containing the daily sea ice extenc for the northern and southern hemispheres, and the climatological means for these two hemispheres
extent = "_seaice_extent_daily_v4.0.csv"
climatology = "_seaice_extent_climatology_1981-2010_v4.0.csv"
# the "climatology" file contains the mean sea ice extent for each day of the year, averaged over the period 1981-2010. Note that this differs from the mean used in the tweet, which is 1991-2020.

if RELOAD_DATA:
    for hemisphere, short in zip(["north", "south"], ["N", "S"]):
        url = (
            "https://noaadata.apps.nsidc.org/NOAA/G02135/" + hemisphere + "/daily/data/"
        )
        filename = url + short + extent
        with open(hemisphere + "_daily.csv", "w") as f:
            f.write(requests.get(filename).text)
        filename = url + short + climatology
        with open(hemisphere + "_climatology.csv", "w") as f:
            f.write(requests.get(filename).text)


tweet_date = datetime(2023, 7, 29)
day_of_tweet = tweet_date.timetuple().tm_yday
today = datetime.now()
```

Now, to this standardization business. A lot of climate data is plotted as "Anomalies", i.e. deviations relative to the "climatological record". That record is usually some mean over previous measurements. The challenge is that, obviously, records don't go back to pre-industrial times. And if they do, they are not reliable. So the average is computed to quite recent trends, even those included in the data itself. In this case, NOAA uses the 30 year period 1981-2010, provided in the `climatology` file:

```python
climatology = pd.read_csv(
    "south_climatology.csv", header=[0], skiprows=[0], skipinitialspace=True
)
climatology
```

```python
# We add this data to our plot:
fig, ax = plt.subplots()

# the data that the climatology is based on:
for year in np.arange(1981, 2011):
    year_data = extent[extent["Year"] == year]
    (avg,) = ax.plot(
        year_data["DayOfYear"],
        year_data["Extent"],
        color=rgb.tue_blue,
        alpha=0.5,
        label="1981-2010",
    )

# and new data:
for year in np.arange(2010, 2026):
    year_data = data_up_to_tweet[data_up_to_tweet["Year"] == year]
    (new,) = ax.plot(
        year_data["DayOfYear"],
        year_data["Extent"],
        color=rgb.tue_red,
        alpha=0.75,
        label="2010-2023",
    )


for i in np.arange(5):
    (pct,) = ax.plot(
        climatology["DOY"],
        climatology.iloc[:, i + 3],
        color=rgb.tue_green,
        linewidth=1,
        label="percentiles",
    )

(mean,) = ax.plot(
    climatology["DOY"],
    climatology["Average Extent"],
    color=rgb.tue_dark,
    linewidth=3,
    label="climatological mean",
)

(std,) = ax.plot(
    climatology["DOY"],
    climatology["Average Extent"] + 2 * climatology["Std Deviation"],
    color=rgb.tue_dark,
    linewidth=1,
    label="climatological std. dev.",
)
ax.plot(
    climatology["DOY"],
    climatology["Average Extent"] - 2 * climatology["Std Deviation"],
    color=rgb.tue_dark,
    linewidth=1,
)

ax.axvline(day_of_tweet, color=rgb.tue_gold, linestyle="-")

ax.set_title("Antarctic sea ice extent, 1979-2023, with climatology")
ax.set_ylabel("Sea ice extent [$10^6$ km$^2$]")
ax.set_xlabel("Day of the year")
ax.set_xlim([0, 364])
ax.xaxis.set_major_locator(mdates.MonthLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter("%b"))
ax.grid(which="major", axis="x")
ax.legend(
    handles=[avg, new, pct, mean, std],
    loc="upper left",
    framealpha=1,
    facecolor="white",
)

plt.savefig("seaice_extent_south_3.pdf")

# and new data:
for year in np.arange(2010, 2026):
    year_data = new_data[new_data["Year"] == year]
    (very_new,) = ax.plot(
        year_data["DayOfYear"],
        year_data["Extent"],
        color=rgb.tue_orange,
        alpha=0.75,
        label="2023-2025",
    )

ax.legend(
    handles=[avg, new, very_new, pct, mean, std],
    loc="upper left",
    framealpha=1,
    facecolor="white",
)

plt.savefig("seaice_extent_south_3_with_new_data.pdf")
```

## The Wald test

now that we have seen where the data comes from, we can do a "signifance level" plot like in the tweet:

```python
fig, ax = plt.subplots()

for year in data_up_to_tweet["Year"].unique():
    year_data = data_up_to_tweet[data_up_to_tweet["Year"] == year]

    # we compute a z-score for each year:
    anomaly = np.array(year_data["Extent"]) - np.array(
        climatology["Average Extent"][year_data["DayOfYear"] - 1]
    )
    std_dev = np.array(climatology["Std Deviation"][year_data["DayOfYear"] - 1])

    if year == 2023:
        color = rgb.tue_red
    else:
        color = rgb.tue_blue

    ax.plot(
        year_data["DayOfYear"],
        anomaly / std_dev,
        color=color,
        alpha=(year - 1978) / (2026 - 1978),
        lw=0.75,
    )

ax.axvline(day_of_tweet, color=rgb.tue_gold, linestyle="-")

ax.set_title(
    "Antarctic Sea Ice extent: daily standard deviations 1989-2023, standard deviations from 1991-2020 climatology",
    fontsize="small",
)
ax.set_xlabel("Day of the year")
ax.text(
    6 * 30 + 10,
    -4.5,
    "2023",
    color=rgb.tue_red,
    fontsize="small",
    ha="center",
    va="top",
)
ax.set_xlim([0, 364])
ax.xaxis.set_major_locator(mdates.MonthLocator(np.arange(2, 13, 1)))
ax.xaxis.set_minor_locator(mdates.MonthLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter("%b"))
ax.grid(which="major", axis="both")
ax.set_ylabel("Standard deviations")

plt.savefig("anomaly.pdf")


for year in new_data["Year"].unique():
    year_data = new_data[new_data["Year"] == year]

    # we compute a z-score for each year:
    anomaly = np.array(year_data["Extent"]) - np.array(
        climatology["Average Extent"][year_data["DayOfYear"] - 1]
    )
    std_dev = np.array(climatology["Std Deviation"][year_data["DayOfYear"] - 1])

    ax.plot(
        year_data["DayOfYear"],
        anomaly / std_dev,
        color=rgb.tue_orange,
        alpha=(year - 1978) / (2026 - 1978),
        lw=0.75,
    )

plt.savefig("anomaly_with_new_data.pdf")
```

Is this data actually Gaussian?

```python
# first, let's have a look at the marginal distribution, i.e. the distribution of the sea ice extent, regardless of the day of the year:
zscores = np.zeros(shape=(366, 30))
climatological_data = extent[extent["Year"] < 2011]

bins = np.linspace(-3, 3, 31)

for day in climatological_data["DayOfYear"].unique():
    day_data = climatological_data[climatological_data["DayOfYear"] == day]
    m = day_data["Extent"].mean()
    s = day_data["Extent"].std()
    hist, b = np.histogram((day_data["Extent"] - m) / s, bins=bins, density=True)
    zscores[day - 1, :] = hist

marginal = np.mean(zscores, axis=0)

fig, axs = plt.subplots(1, 3)
ax = axs[0]
ax.matshow(
    zscores.T, cmap="Blues", extent=[0, 365, -3, 3], aspect="auto", origin="lower"
)
ax.xaxis.tick_bottom()
ax.set_xlabel("Day of the year")
ax.set_ylabel("$z$ score")
ax.set_title("standardized sea ice extent")

ax = axs[1]
ax.plot(
    bins[:-1] + 0.5 * (bins[1] - bins[0]),
    marginal.cumsum() / marginal.sum(),
    ".-",
    color=rgb.tue_blue,
    ms=3,
    label="climatological mean",
)
ax.plot(
    bins[:-1] + 0.5 * (bins[1] - bins[0]),
    norm.cdf(bins[:-1]),
    ".-",
    color=rgb.tue_dark,
    ms=3,
    label="normal distribution",
)
ax.set_xlabel("$z$ score")
ax.xaxis.set_major_locator(plt.MultipleLocator(1))
ax.xaxis.set_minor_locator(plt.FixedLocator(bins))
ax.grid(which="both", axis="x")
ax.set_xlim([-3, 3])
ax.legend(loc="upper left", framealpha=1, facecolor="white")
ax.set_ylabel("cumulative probability")
ax.set_title("marginal cdf")

ax = axs[2]
alpha_score = norm.cdf(bins[:-1] + 0.5 * (bins[1] - bins[0]))
ax.plot(alpha_score, marginal.cumsum() / marginal.sum(), ".-", color=rgb.tue_blue, ms=3)
ax.plot([0, 1], [0, 1], color=rgb.tue_dark, lw=0.5)
ax.set_xlabel("normal cdf")
ax.set_ylabel("empirical cdf")
ax.grid(which="both", axis="both")
ax.set_xlim([0, 1])
ax.set_ylim([0, 1])
ax.xaxis.set_major_locator(plt.MultipleLocator(0.2))
ax.yaxis.set_major_locator(plt.MultipleLocator(0.2))
ax.set_title("QQ plot")

plt.savefig("qq.pdf")
```

```python
from scipy.stats import kstest

# the empirical cdfs:
climatological_data = extent[extent["Year"] < 2011]
new_data = extent[extent["Year"] >= 2011]
zscores = []
zscores_new = []

for day in range(1, 366):
    day_data = climatological_data[climatological_data["DayOfYear"] == day]
    m = day_data["Extent"].mean()
    s = day_data["Extent"].std()
    for year in range(1981, 2011):
        year_data = day_data[day_data["Year"] == year]["Extent"].values
        if len(year_data) > 0:
            zscores.append((year_data[0] - m) / s)
    for year in range(2011, 2026):
        year_data = new_data[new_data["DayOfYear"] == day]["Extent"].values
        if len(year_data) > 0:
            zscores_new.append((year_data[0] - m) / s)


zscores = np.array(zscores)
zscores.sort()

zscores_new = np.array(zscores_new)
zscores_new.sort()

marginal_cdf = np.arange(len(zscores)) / len(zscores)
marginal_cdf_new = np.arange(len(zscores_new)) / len(zscores_new)

norm_cdf = norm.cdf(zscores)

ks_stat1, ks_p1 = kstest(zscores, norm.cdf, alternative="two-sided", method="exact")
print(ks_stat1, ks_p1)

ks_stat2, ks_p2 = kstest(zscores, zscores_new, alternative="two-sided", method="exact")
print(ks_stat2, ks_p2)

# and compute the KS statistic:
my_ks_stat = np.max(np.abs(marginal_cdf - norm_cdf))
my_ks_loc = np.argmax(np.abs(marginal_cdf - norm_cdf))

def empirical_cdf(x):
    return np.mean(zscores[:,None] <= x[None,:], axis=0)

my_ks_stat2 = np.max(np.abs(marginal_cdf_new - empirical_cdf(zscores_new)))
my_ks_loc2 = np.argmax(np.abs(marginal_cdf_new - empirical_cdf(zscores_new)))

print(my_ks_stat, my_ks_loc)
```

```python
plt.rcParams.update(bundles.beamer_moml(rel_height=0.5))
fig, ax = plt.subplots()
ax.plot(zscores, marginal_cdf, "-", color=rgb.tue_blue, label="climatological mean")
ax.plot(zscores, norm_cdf, "-", color=rgb.tue_dark, label="normal distribution")

ax.axvline(zscores[my_ks_loc], color=rgb.tue_darkgreen, linestyle="-")
ax.text(
    zscores[my_ks_loc] + 0.1,
    0.5,
    r"$\sqrt{n} D= \sqrt{n}\cdot \sup_x \left| F_n(x) - F_0(x) \right| ="
    + f"{np.sqrt(len(zscores)) * my_ks_stat:.2f}$",
    color=rgb.tue_blue,
    fontsize="small",
    va="top",
    ha="left",
    bbox=dict(facecolor="white", alpha=0.9, edgecolor="none", pad=1),
)
ax.text(
    zscores[my_ks_loc] + 0.1,
    0.2,
    r"$p = " + f"{ks_p:.2e}$",
    color=rgb.tue_blue,
    fontsize="small",
    va="top",
    ha="left",
    bbox=dict(facecolor="white", alpha=0.9, edgecolor="none", pad=1),
)

ax.set_xlabel("$z$ score")
ax.xaxis.set_major_locator(plt.MultipleLocator(1))
ax.xaxis.set_minor_locator(plt.FixedLocator(bins))
ax.grid(which="both", axis="x")
ax.set_xlim([-3, 3])
ax.legend(loc="upper left", framealpha=1, facecolor="white")
ax.set_ylabel("cumulative probability")
# ax.set_title("marginal cdf")
plt.savefig("KS_one_sample.pdf")
```

```python
plt.rcParams.update(bundles.beamer_moml(rel_height=0.5))
fig, ax = plt.subplots()
ax.plot(zscores, marginal_cdf, "-", color=rgb.tue_blue, label="climatological mean")
ax.plot(zscores, norm_cdf, "-", color=rgb.tue_dark, label="normal distribution")

ax.plot(zscores_new, marginal_cdf_new, "-", color=rgb.tue_orange, label="Years 2011-2025")

ax.axvline(zscores[my_ks_loc], color=rgb.tue_darkgreen, linestyle="-")
ax.text(
    zscores[my_ks_loc] + 0.1,
    0.5,
    r"$\sqrt{n} D= \sqrt{n}\cdot \sup_x \left| F_n(x) - F_0(x) \right| ="
    + f"{np.sqrt(len(zscores)) * my_ks_stat:.2f}$",
    color=rgb.tue_blue,
    fontsize="small",
    va="top",
    ha="left",
    bbox=dict(facecolor="white", alpha=0.9, edgecolor="none", pad=1),
)
ax.text(
    zscores[my_ks_loc] + 0.1,
    0.2,
    r"$p = " + f"{ks_p:.2e}$",
    color=rgb.tue_blue,
    fontsize="small",
    va="top",
    ha="left",
    bbox=dict(facecolor="white", alpha=0.9, edgecolor="none", pad=1),
)

ax.axvline(zscores_new[my_ks_loc2], color=rgb.tue_orange, linestyle="-")
ax.text(
    zscores_new[my_ks_loc2] - 0.1,
    0.55,
    r"$\sqrt{n} D= \sqrt{n}\cdot \sup_x \left| F_n(x) - F_0(x) \right| ="
    + f"{np.sqrt(len(zscores_new)) * my_ks_stat2:.2f}$",
    color=rgb.tue_orange,
    fontsize="small",
    va="top",
    ha="right",
    bbox=dict(facecolor="white", alpha=0.9, edgecolor="none", pad=1),
)
ax.text(
    -2,
    0.3,
    r"$p = " + f"{ks_p2:.2e}$",
    color=rgb.tue_orange,
    fontsize="small",
    va="top",
    ha="center",
    bbox=dict(facecolor="white", alpha=0.9, edgecolor="none", pad=1),
)



ax.set_xlabel("$z$ score")
ax.xaxis.set_major_locator(plt.MultipleLocator(1))
ax.xaxis.set_minor_locator(plt.FixedLocator(bins))
ax.grid(which="both", axis="x")
ax.set_xlim([-3, 3])
ax.legend(loc="upper left", framealpha=1, facecolor="white")
ax.set_ylabel("cumulative probability")
# ax.set_title("marginal cdf")
plt.savefig("KS.pdf")
```

Some things to note:

* Of course, the downward deviation in 2023 really stands out. 
* We also see that, since the tweet, the deviation to "minus 7 sigma" has largely recovered, and we are now at -4. Still not good, but it's coming back up. 
* It also *seems* like the *overall* variation (both up and down) is increasing over time?
* A less obvious thing: It seems that there's a lot more dynamic happening around June to October (i.e. the Antarctic Winter, when the extent is largest)

Thus, here are a few things to do:

* First, we'd like to confirm our hypothesis that variance has increased over time. Doing so would invalidate the whole "this is a six-sigma event" narrative.
* We'd also like to know what's happening here. One simple first thing to do is just to compare with the northern hemisphere. The other one is to make a visual plot of the sea ice itself, too see where that sudden drop in July this year came from. 

First, let's check what the distributions look like per year

```python
plt.rcParams.update(bundles.beamer_moml())
fig, axs = plt.subplots(1, 2)

for year in np.arange(1980, 2026):
    year_data = extent[extent["Year"] == year]
    anomaly = np.array(year_data["Extent"]) - np.array(
        climatology["Average Extent"][year_data["DayOfYear"] - 1]
    )
    # hist = np.histogram(anomaly, bins=30, range=(-3,3), density=True)
    kde = stats.gaussian_kde(anomaly)

    # axs[0].plot(hist[1][:-1], hist[0], color=rgb.tue_blue, alpha=(year - 1980) / (2023 - 1980))
    xp = np.linspace(-3, 3, 200)

    if year == 2023:
        color = rgb.tue_red
    elif year > 2023:
        color = rgb.tue_orange
    else:
        color = rgb.tue_blue

    axs[0].plot(xp, kde.pdf(xp), color=color, alpha=(year - 1980) / (2026 - 1980))
    # axs[0].fill_between(xp, kde.pdf(xp), color=rgb.tue_blue, alpha=(year - 1980) / (2023 - 1980))

    axs[1].plot(
        [year, year],
        [np.mean(anomaly) - np.std(anomaly), np.mean(anomaly) + np.std(anomaly)],
        color=color,
        alpha=(year - 1980) / (2026 - 1980),
    )
    parts = axs[1].violinplot(
        anomaly,
        positions=[year],
        showmeans=True,
        showmedians=True,
        showextrema=True,
        widths=1.5,
    )
    for pc in parts["bodies"]:
        pc.set_facecolor(color)
        pc.set_edgecolor(color)
        pc.set_alpha((year - 1980) / (2030 - 1980))
    for pc in [
        parts["cbars"],
        parts["cmins"],
        parts["cmaxes"],
        parts["cmeans"],
        parts["cmedians"],
    ]:
        pc.set_color(color)
        pc.set_alpha((year - 1980) / (2026 - 1980))

axs[0].set_xlabel("Sea ice anomaly [$10^6$ km$^2$]")
axs[0].set_ylabel("kernel density estimate")
axs[0].set_xlim([-3, 3])
axs[0].grid(which="major", axis="x")

axs[1].set_xlabel("Year")
axs[1].set_ylabel("Sea ice anomaly [$10^6$ km$^2$]")
axs[1].xaxis.set_major_locator(plt.MultipleLocator(5))
axs[1].xaxis.set_minor_locator(plt.MultipleLocator(1))
axs[1].grid(which="major", axis="both")
axs[1].set_xlim([1980, 2026])

axs[1].axvspan(1981, 2010, color=rgb.tue_gray, alpha=0.5)
axs[1].text(
    1995.5,
    -2.0,
    "reference period (1981-2010)",
    color=rgb.tue_dark,
    fontsize=8,
    ha="center",
    va="center",
    rotation=0,
)
axs[1].axhline(0, color=rgb.tue_dark, linewidth=0.5)

plt.savefig("anomaly_distribution.pdf")
```

So this is interesting, as we see that the truly wierd behaviour in the anomaly probably didn't start this year, but in 2016. Before that, we see a *buildup* of ice, actually.

```python
import matplotlib.pyplot as plt
import numpy as np

# Convert the date to day of the year
extent["DayOfYear"] = extent["Date"].dt.dayofyear
extent["Year"] = extent["Date"].dt.year

# Create a figure and a polar axis
fig = plt.figure()
ax = fig.add_subplot(111, polar=True)

# Plot each year's data as one full circle
for y in extent["Year"].unique():
    # Select the data for this year
    year_data = extent[extent["Year"] == y]

    # Convert the day of the year to radians
    theta = 2 * np.pi * year_data["DayOfYear"] / 365

    if y == 2023:
        color = rgb.tue_red
    elif y > 2023:
        color = rgb.tue_orange
    else:
        color = rgb.tue_blue
    # Plot the data
    ax.plot(
        theta,
        year_data["Extent"],
        label=y,
        color=color,
        alpha=(y - 1978) / (2026 - 1978),
    )

# Set the direction of the plot to clockwise
ax.set_theta_direction(-1)

# Set the start of the plot to January
ax.set_theta_offset(np.pi / 2)

# Set the angle coordinates to show months of the year
ax.set_xticks(np.linspace(0, 2 * np.pi, 12, endpoint=False))
ax.set_xticklabels(
    ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
)

ax.set_title("Northern hemisphere sea ice extent")
ax.yaxis.set_major_locator(plt.MultipleLocator(5))
ax.set_ylim(0, 20)
plt.savefig("seaice_extent_north_1_polar.pdf")
```

```python
# We add this data to our plot:
fig, ax = plt.subplots()

# the data that the climatology is based on:
for year in np.arange(1981, 2011):
    year_data = extent[extent["Year"] == year]
    (avg,) = ax.plot(
        year_data["DayOfYear"],
        year_data["Extent"],
        color=rgb.tue_blue,
        alpha=0.5,
        label="1981-2010",
    )

# and new data:
for year in np.arange(2010, 2026):
    year_data = extent[extent["Year"] == year]
    (new,) = ax.plot(
        year_data["DayOfYear"],
        year_data["Extent"],
        color=rgb.tue_orange,
        alpha=0.75,
        label="2010-2025",
    )

for i in np.arange(5):
    (pct,) = ax.plot(
        climatology["DOY"],
        climatology.iloc[:, i + 3],
        color=rgb.tue_green,
        linewidth=1,
        label="percentiles",
    )

(mean,) = ax.plot(
    climatology["DOY"],
    climatology["Average Extent"],
    color=rgb.tue_red,
    linewidth=3,
    label="climatological mean",
)

(std,) = ax.plot(
    climatology["DOY"],
    climatology["Average Extent"] + 2 * climatology["Std Deviation"],
    color=rgb.tue_red,
    linewidth=1,
    label="climatological std. dev.",
)
ax.plot(
    climatology["DOY"],
    climatology["Average Extent"] - 2 * climatology["Std Deviation"],
    color=rgb.tue_red,
    linewidth=1,
)

ax.axvline(day_of_tweet, color=rgb.tue_gold, linestyle="-")

ax.set_ylabel("Sea ice extent [$10^6$ km$^2$]")
ax.set_xlabel("Day of the year")
ax.set_xlim([0, 364])
ax.xaxis.set_major_locator(mdates.MonthLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter("%b"))
ax.grid(which="major", axis="x")
ax.legend(
    handles=[avg, new, pct, mean, std],
    loc="lower left",
    framealpha=1,
    facecolor="white",
)

plt.savefig("seaice_extent_north_3.pdf")
```

```python
fig, ax = plt.subplots()

for year in extent["Year"].unique():
    year_data = extent[extent["Year"] == year]

    # we compute a z-score for each year:
    anomaly = np.array(year_data["Extent"]) - np.array(
        climatology["Average Extent"][year_data["DayOfYear"] - 1]
    )
    std_dev = np.array(climatology["Std Deviation"][year_data["DayOfYear"] - 1])

    if year == 2023:
        color = rgb.tue_red
    elif year > 2023:
        color = rgb.tue_orange
    else:
        color = rgb.tue_blue

    ax.plot(
        year_data["DayOfYear"],
        anomaly / std_dev,
        color=color,
        alpha=(year - 1978) / (2026 - 1978),
        lw=0.75,
    )

ax.axvline(day_of_tweet, color=rgb.tue_gold, linestyle="-")

ax.set_title(
    "Arctic Sea Ice extent: daily standard deviations 1989-2023. standard deviations from 1991-2020 climatology",
    fontsize="small",
)
ax.set_xlabel("Day of the year")
ax.text(
    6 * 30 + 10,
    -4.5,
    "2023",
    color=rgb.tue_red,
    fontsize="small",
    ha="center",
    va="top",
)
ax.set_xlim([0, 364])
ax.xaxis.set_major_locator(mdates.MonthLocator(np.arange(2, 13, 1)))
ax.xaxis.set_minor_locator(mdates.MonthLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter("%b"))
ax.grid(which="major", axis="both")
ax.set_ylabel("Standard deviations")

plt.savefig("anomaly_north.pdf")
```

```python
from scipy.stats import norm

fig, ax = plt.subplots()

for year in extent["Year"].unique():
    year_data = extent[extent["Year"] == year]

    # we compute a z-score for each year:
    anomaly = np.array(year_data["Extent"]) - np.array(
        climatology["Average Extent"][year_data["DayOfYear"] - 1]
    )
    std_dev = np.array(climatology["Std Deviation"][year_data["DayOfYear"] - 1])

    if year == 2023:
        color = rgb.tue_red
    elif year > 2023:
        color = rgb.tue_orange
    else:
        color = rgb.tue_blue

    z_score = norm.cdf(-np.abs(anomaly / std_dev))

    ax.plot(
        year_data["DayOfYear"],
        z_score,
        color=color,
        alpha=(year - 1978) / (2026 - 1978),
        lw=0.75,
    )

ax.axvline(day_of_tweet, color=rgb.tue_gold, linestyle="-")

ax.set_yscale("log")
ax.set_ylim([1e-7, 1])

ax.set_title(
    "Arctic Sea Ice extent: daily standard deviations 1989-2023. standard deviations from 1991-2020 climatology",
    fontsize="small",
)
ax.set_xlabel("Day of the year")
ax.set_xlim([0, 364])
ax.xaxis.set_major_locator(mdates.MonthLocator(np.arange(2, 13, 1)))
ax.xaxis.set_minor_locator(mdates.MonthLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter("%b"))
ax.grid(which="major", axis="both")
ax.set_ylabel("Cumulative probability")

plt.savefig("anomaly_north_p_value.pdf")
```
